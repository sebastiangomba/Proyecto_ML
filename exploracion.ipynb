```python
#cargar las librerias 

import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd
import seaborn as sns
import openpyxl as pxl
```

### Empezamos la limpieza del dataset

```python
#cargamos el dataset
data = pd.read_excel('Dataset/data_academic_performance.xlsx')  
data.head()
```

```python
data.info()
data.describe()
```

```python
#ver columnas con Nans
data.isnull().sum()
```

```python
#eliminamos columna unmamed: 9 (si existe)
if 'Unnamed: 9' in data.columns:
    data = data.drop(columns=['Unnamed: 9'])

data.isnull().sum()

# Eliminar variables irrelevantes
data = data.drop(['SYSBEND', 'TV', 'DVD'], errors='ignore')
```

```python
data["REVENUE"]
```

```python
data = data.dropna(subset=["REVENUE"])
data = data[data["REVENUE"] != 0]
data.isnull().sum()
```

```python
data['REVENUE'].value_counts()
```

```python
# Definir un mapeo
orden = {'less than 1 LMMW': 1, 'Between 1 and less than 2 LMMW': 2, 'Between 2 and less than 3 LMMW': 3, 'Between 3 and less than 5 LMMW': 4, 'Between 5 and less than 7 LMMW': 5, 'Between 7 and less than 10 LMMW': 6, '10 or more LMMW': 7}

# Aplicar el mapeo personalizado
data['REVENUE'] = data['REVENUE'].map(orden)
```

```python
data['REVENUE'].value_counts()
```

### Antes de continuar con la limpieza, hacemos graficos para el analisis exploratorio

```python
fig, ax = plt.subplots(figsize=(6,4))
ax.boxplot(data["G_SC"].dropna(), vert=True, patch_artist=False, showfliers=False)
ax.set_title("Dispersión del Puntaje Global (G_SC)")
ax.set_ylabel("Puntaje G_SC")
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
cols_pro = ['QR_PRO', 'CR_PRO', 'CC_PRO', 'ENG_PRO', 'WC_PRO', 'G_SC']

means_pc = data.groupby("COMPUTER")[cols_pro].mean().round(1)
means_pc.plot(kind="bar", figsize=(8,4))
plt.title("Promedio de Puntajes SABER-PRO según acceso a computador")
plt.ylabel("Puntaje promedio")
plt.xlabel("¿Tiene computador?")
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

pct_computer = (data["COMPUTER"].value_counts(normalize=True) * 100).round(1)
print("Porcentaje con computador (%)")
print(pct_computer)
```

```python
cols_pro = ['QR_PRO', 'CR_PRO', 'CC_PRO', 'ENG_PRO', 'WC_PRO', 'G_SC']

# Agrupar y calcular medias por valor numérico de REVENUE
means_rev = data.groupby("REVENUE")[cols_pro].mean().round(1)

# Gráfico tipo línea (más apropiado si ahora es numérico y ordenado)
means_rev.plot(kind="line", marker='o', figsize=(8,4))
plt.title("Promedio de Puntajes SABER-PRO según nivel de ingresos (REVENUE)")
plt.ylabel("Puntaje promedio")
plt.xlabel("Nivel de ingreso (1=menor, 7=mayor)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
cols_pro = ['QR_PRO', 'CR_PRO', 'CC_PRO', 'ENG_PRO', 'WC_PRO', 'G_SC']

# Educación del padre
edu_father = data.groupby("EDU_FATHER")[cols_pro].mean().round(1).sort_values("G_SC", ascending=False)
fig, ax = plt.subplots(figsize=(10,4))
edu_father["G_SC"].plot(kind="bar", ax=ax)
ax.set_title("G_SC promedio según educación del padre")
ax.set_ylabel("Puntaje promedio (SABER-PRO)")
ax.set_xlabel("Nivel educativo del padre")
plt.xticks(rotation=45, ha="right")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# Educación de la madre
edu_mother = data.groupby("EDU_MOTHER")[cols_pro].mean().round(1).sort_values("G_SC", ascending=False)
fig, ax = plt.subplots(figsize=(10,4))
edu_mother["G_SC"].plot(kind="bar", ax=ax)
ax.set_title("G_SC promedio según educación de la madre")
ax.set_ylabel("Puntaje promedio (SABER-PRO)")
ax.set_xlabel("Nivel educativo de la madre")
plt.xticks(rotation=45, ha="right")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
use_cols = ['MAT_S11','CR_S11','CC_S11','BIO_S11','ENG_S11','QR_PRO','CR_PRO','CC_PRO','ENG_PRO','WC_PRO','G_SC']
use_cols = [c for c in use_cols if c in data.columns]
corr = data[use_cols].corr(numeric_only=True)

fig, ax = plt.subplots(figsize=(9,7))
im = ax.imshow(corr, vmin=-1, vmax=1)
ax.set_xticks(range(len(use_cols))); ax.set_xticklabels(use_cols, rotation=90)
ax.set_yticks(range(len(use_cols))); ax.set_yticklabels(use_cols)
ax.set_title("Matriz de correlación entre componentes SABER-11 y SABER-PRO")
plt.colorbar(im, ax=ax, fraction=0.045, pad=0.04)
plt.tight_layout()
plt.show()
```

```python
top_univ = data.groupby("UNIVERSITY")["G_SC"].mean().sort_values(ascending=False).head(15)
top_univ.plot(kind="bar", figsize=(10,4))
plt.title("Top 15 Universidades por puntaje promedio (G_SC)")
plt.ylabel("Puntaje promedio G_SC")
plt.xlabel("Universidad")
plt.xticks(rotation=45, ha="right")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
if "GENDER" in data.columns and "QUARTILE" in data.columns:
    ct = pd.crosstab(data["GENDER"], data["QUARTILE"], normalize="index") * 100
    ax = ct.plot(kind="bar", stacked=True, figsize=(7,4))
    plt.title("Distribución porcentual de cuartiles (QUARTILE) por género")
    plt.ylabel("% de estudiantes")
    plt.xlabel("Género")
    plt.legend(title="Cuartil", bbox_to_anchor=(1.05, 1), loc="upper left")
    plt.tight_layout()
    plt.show()
else:
    print("Faltan columnas GENDER o QUARTILE.")
```

### Finalizamos con la limpieza de datos y preparación del dataset

```python
#Variables Binarias 

binary_map = {
    'Yes': 1, 'No': 0,
    'F': 0, 'M': 1,
    'private': 1, 'public': 0
}
for col in ['GENDER', 'INTERNET', 'COMPUTER', 'SCHOOL_NAT']:
    if col in data.columns:
        data[col] = data[col].map(binary_map).fillna(data[col])

data.head()
```

```python
data["PEOPLE_HOUSE"].value_counts()
```

```python
# Estrato de "Stratum 1" a "1" y asi
data['STRATUM'] = data['STRATUM'].astype(str).str.extract('(\d+)')[0].astype(float)

# convertir texto a números de personas en la casa
map_people = {
    'One': 1, 'Two': 2, 'Three': 3, 'Four': 4, 'Five': 5,
    'Six': 6, 'Seven': 7, 'Eight': 8, 'Ten': 10,
    'Once': 11, 'Nueve': 9, 'Twelve or more': 12
}

data['PEOPLE_HOUSE'] = (
    data['PEOPLE_HOUSE']
    .map(map_people)
)


data['PEOPLE_HOUSE']
```

```python
data['STRATUM']
```

```python
#separamos columnas categoricas y numericas
data_categoricas = data.select_dtypes(include=['object'])
data_numericas = data.select_dtypes(include=['number'])

data_numericas.head()
```

```python
# Calcular matriz de correlación
corr = data_numericas.corr().round(2)

# Mostrar valores de correlación más altos (opcional)
print("Top correlaciones:")
corr_pairs = (
    corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
    .stack()
    .sort_values(ascending=False)
)
print(corr_pairs.head(15))

# Graficar heatmap simple
plt.figure(figsize=(12, 10))
plt.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar(label="Coeficiente de correlación")
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Matriz de Correlación de Variables Numéricas", pad=20)
plt.tight_layout()
plt.show()
```

### Finalmente eliminamos todas las columnas que por obvios motivos no se van a incluir para el modelo, 

```python

cols_to_drop = ['QR_PRO', 'CR_PRO', 'CC_PRO', 'ENG_PRO', 'WC_PRO', 'FEP_PRO','PERCENTILE']


data = data.drop(columns=[c for c in cols_to_drop if c in data.columns], errors='ignore')
```
