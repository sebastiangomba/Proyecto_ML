```python
# Definir un mapeo
orden = {'less than 1 LMMW': 1, 'Between 1 and less than 2 LMMW': 2, 'Between 2 and less than 3 LMMW': 3, 'Between 3 and less than 5 LMMW': 4, 'Between 5 and less than 7 LMMW': 5, 'Between 7 and less than 10 LMMW': 6, '10 or more LMMW': 7}

# Aplicar el mapeo personalizado
data['REVENUE'] = data['REVENUE'].map(orden)
```
